```python
import seaborn as sns
```

```python
draw = sns.histplot(data=df,x="categoria",hue="categoria",kde=True, bins=5)
draw.set_title("Distribuição por categoria")
draw.set_xlabel("Categoria")
draw.set_ylabel("Vendas")
```

```python
draw = sns.boxplot(data=df,x="categoria",y="preco_unitario")
draw.set_title("Preço por Categoria")
draw.set_xlabel("Categoria")
draw.set_ylabel("Preços")
```
